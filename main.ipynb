```python
# Import required libraries for database, data handling, and GUI
import sqlite3
import pandas as pd
import tkinter as tk
from tkinter import messagebox


# --- Database Section ---
# This function creates the database and table if they do not exist
def setup_database():
    # Connect to the SQLite database file (creates it if missing)
    conn = sqlite3.connect("data.db")
    c = conn.cursor()
    
    # Create a table named hotel with columns: ID, Name, Cost, and Weight
    c.execute('''
           CREATE TABLE IF NOT EXISTS hotel (
        FIND INTEGER PRIMARY KEY AUTOINCREMENT,
        FNAME TEXT NOT NULL,
        COST INTEGER NOT NULL,
        WEIGHT INTEGER
    )
    ''')
    conn.commit()
    conn.close()




# --- Button Click Function ---
# This function runs when the user clicks the Save button
def submit_data():
    # Get values from the input text boxes
    na = entry_name.get()
    co = entry_cost.get()
    wi = entry_weight.get()

    # Check if name or cost fields are empty
    if not na or not co :
        messagebox.showerror("Error", "Please fill all required fields")
        return

    try:
        # Convert cost and weight to integer numbers
        co = int(co)
        wi = int(wi)
        
        # Connect to database and insert the new record
        conn = sqlite3.connect("data.db")
        c = conn.cursor()
        c.execute("INSERT INTO hotel (FNAME, COST, WEIGHT) VALUES (?, ?, ?)", (na, co, wi))
        conn.commit()
        conn.close()
        
        # Show success message and clear all input fields
        messagebox.showinfo("Success", "Data saved successfully")
        entry_name.delete(0, tk.END)
        entry_cost.delete(0, tk.END)
        entry_weight.delete(0, tk.END)
        
        
    except ValueError:
        # Show error if cost or weight are not valid numbers
        messagebox.showerror("Error", "Cost and Weight must be numbers")

# --- Graphical User Interface (GUI) Section ---

# Step 1: Setup the database (create table if not exists)
setup_database()

# Step 2: Create the main application window
root = tk.Tk()
root.title("Hotel Registration Form")
root.geometry("300x250")

# Step 3: Create labels and input boxes for user data

# Name input field
tk.Label(root, text="Name:").pack(pady=5)
entry_name = tk.Entry(root)
entry_name.pack(pady=5)

# Cost input field
tk.Label(root, text="Cost:").pack(pady=5)
entry_cost = tk.Entry(root)
entry_cost.pack(pady=5)

# Weight input field (default value is 0)
tk.Label(root, text="Weight:").pack(pady=5)
entry_weight = tk.Entry(root)
entry_weight.pack(pady=5)
entry_weight.insert(0,"0")

# Step 4: Create the Save button that calls submit_data function
btn_save = tk.Button(root, text="Save to Database", command=submit_data, bg="green", fg="white")
btn_save.pack(pady=20)

# Start the main event loop to run the application
root.mainloop()
```
